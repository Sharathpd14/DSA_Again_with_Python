```python
n = 1

while n<=10:  # condition for n
    print("log n time complexity")
    n*=2
```

```python
n = 5  # O(1)

for i in range(0,n):    # iterates n times
    print(i, end = " ")         # prints ith value n times 
    
print("\n")
print(f"Time complexity is O(n)")
```

```python
n = 3
m = 3

for i in range(n):     
    for j in range(m):
        print(i+j, end = " ")
        
print("\n")
print("Time complexity is O(n^2)")
```
