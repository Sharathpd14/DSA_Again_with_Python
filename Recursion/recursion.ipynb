```python
def add(n):
    if n == 1:
        return 1
    return add(n-1) + n
```

```python
add(10)
```

```python
def fact(n):
    if n == 1 or n == 0:
        return 1
    return n * fact(n-1)
```

```python
fact(0)
```

```python
num = 5
fact_variable = 1
```

```python
for i in range(1,num+1):
    fact_variable *= i 
```

```python
fact_variable
```

```python
def fib(n):
    if n == 1 or n==2:
        return 1
    return fib(n-1) + fib(n-2)
    
    
```

```python
fib(7)
```

```python
def print_fib(n, a=1, b=1):
    if n == 0:
        return
    print(a)
    print_fib(n-1, b, a+b)
```

```python
print_fib(7)
```
