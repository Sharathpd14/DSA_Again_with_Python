```python
class Node:
    def __init__(self, data, next_node_address = None):
        self.data = data
        self.next_node_address = next_node_address
        
```

```python
class SinglyLinkedList:
    def __init__(self, head = None):
        self.head = head
        
    def insertAtStart(self,value):
        temp = Node(value)
        temp.next_node_address = self.head
        self.head = temp
    
    def insertAtMidle(self,value,location_value):
        temp = Node(value)
        first_node_pointer = self.head
        
        while (first_node_pointer.next_node_address != None):
            if first_node_pointer.data == location_value:
                temp.next_node_address = first_node_pointer.next_node_address
                first_node_pointer.next_node_address = temp
            first_node_pointer = first_node_pointer.next_node_address
        
    
    def insertAtEnd(self,value):
        temp = Node(value)
        
        if self.head != None:
            assistant_temp = self.head
            while assistant_temp.next_node_address !=None:
                assistant_temp = assistant_temp.next_node_address
            assistant_temp.next_node_address = temp
            
        else:
            self.head = temp
            
    
    def deletedLinkedList(self, value):
        node_pointer = self.head
        prev_node_pointer = node_pointer
        
        if node_pointer.data == value :
            self.head = node_pointer.next_node_address
            
        
        
        while node_pointer.next_node_address != None:
            if node_pointer.data == value:
                prev_node_pointer.next_node_address = node_pointer.next_node_address
                break
            else:
                prev_node_pointer = node_pointer
                node_pointer = node_pointer.next_node_address
            
        if node_pointer.data == value:
            prev_node_pointer.next_node_address = None
    
    
            
    def printLinkedList(self):
        assistant_temp = self.head
        while assistant_temp.next_node_address !=None:
            print(assistant_temp.data, end =" ")
            assistant_temp = assistant_temp.next_node_address
        print(assistant_temp.data, end = " ")
        
```

```python
obj_1 = SinglyLinkedList()

obj_1.insertAtEnd(16)
obj_1.insertAtEnd(10)
obj_1.insertAtEnd(20)
obj_1.insertAtEnd(30)
obj_1.insertAtEnd(40)
obj_1.insertAtEnd(21)
obj_1.insertAtStart(6)
obj_1.insertAtMidle(37,20)
obj_1.deletedLinkedList(6)
```
