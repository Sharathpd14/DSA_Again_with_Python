```python
class Queue:
    def __init__(self):
        self.items = []
        
    def isEmpty(self):
        return len(self.items) == 0
    
    def insert(self,value):
        self.items.append(value)
        
    def delete(self):
        if self.isEmpty():
            print("Queue is empty")
        else:
            return self.items.pop(0)
```

```python
Q1 = Queue()
```

```python
Q1.isEmpty()
```

```python
Q1.insert(10)
Q1.insert(20)
Q1.insert(30)
```

```python
Q1.isEmpty()
```

```python
print(Q1.delete())
print(Q1.delete())
```

```python
class Dequeue:
    def __init__(self):
        self.items = []
        
    def isEmpty(self):
        return len(self.items) == 0
    
    def insertAtStart(self,value):
        self.items.insert(0,value)
    
    def insertAtEnd(self,value):
        self.items.append(value)
    
    def deleteAtStart(self):
        if self.isEmpty():
            print("Queue is empty")
        else:
            return self.items.pop(0)
    
    def deleteAtEnd(self):
        if self.isEmpty():
            print("Queue is empty")
        else:
            return self.items.pop()
        
    def returnElements(self):
        return self.items
```

```python
Q2 = Dequeue()
```

```python
Q2.insertAtEnd(10)
print(Q2.returnElements())
Q2.insertAtEnd(20)
print(Q2.returnElements())
Q2.insertAtEnd(30)
print(Q2.returnElements())
Q2.insertAtStart(40)
print(Q2.returnElements())
Q2.deleteAtEnd()
print(Q2.returnElements())
Q2.deleteAtStart()
print(Q2.returnElements())
```
