```python
class CircularQueue:
    def __init__(self, size):
        self.size = size
        self.items = [None] * size
        self.front = self.rear = -1
        
    def enqueue(self,value):
        if (self.rear + 1) % self.size == self.front :
            print("Queue is full")
            
        elif self.front == -1:
            self.front = self.rear = 0
            self.items[self.rear] = value
            
        else:
            self.rear = (self.rear + 1) % self.size
            self.items[self.rear] = value
            
    def dequeue(self):
        if self.front == -1:
            print("Queue is empty")
        elif self.front == self.rear:
            print(f"{self.items[self.front]} is deleted. ")
            self.front = self.rear = -1
        else:
            print(f"{self.items[self.front]} is deleted. ")
            self.front = (self.front+1) % self.size
            
            
    
```

```python
C_Queue1 = CircularQueue(5) 
```

```python
def add(n):
    if n == 1:
        return 1
    return add(n-1) + n
```

```python
add(10)
```

```python
def fact(n):
    if n == 1 or n == 0:
        return 1
    return n * fact(n-1)
```

```python
fact(0)
```

```python
def fib(n):
    if n == 1 or n==2:
        return 1
    return fib(n-1) + fib(n-2)
    
    
```
