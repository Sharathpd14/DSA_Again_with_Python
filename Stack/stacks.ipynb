```python
class Stack:
    def __init__(self):
        self.arr = []
        
    def length(self):
        return len(self.arr)
        
    def push(self,value):
        self.arr.insert(0,value)
        print(f"{value} is added to your stack")
    
    def peek(self):
        if len(self.arr) == 0:
            raise Exception("Stack is empty")
        else:
            return f"The top element of your stack is {self.arr[0]}"
    
   
    
    def pop(self):
        if len(self.arr) == 0:
            raise Exception("Stack is empty")
        else:
            return f"The deleted item is {self.arr.pop(0)}"
```

```python
myStack = Stack()
```

```python
myStack.push(10)
```

```python
myStack.peek()
```

```python
myStack.push(32)
```

```python
myStack.push(34)
myStack.push(3)
```

```python
myStack.pop()
```

```python
myStack.peek()
```
