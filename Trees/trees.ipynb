```python
## Traversing
```

```python
class TreeNode:
    def __init__(self,value):
        self.left = None
        self.right = None
        self.data = value
```

```python
root = TreeNode(1)
```

```python
root.left = TreeNode(2)
root.right = TreeNode(5)
```

```python
root.left.left = TreeNode(3)
root.left.right = TreeNode(4)
```

```python
root.right.left = TreeNode(6)
root.right.right = TreeNode(7)
```

```python
def preorder(root):
        if root != None:
            print(root.data, end = " ")
            preorder(root.left)
            preorder(root.right)
```

```python
preorder(root)
```

```python
def ineorder(root):
        if root != None:
            preorder(root.left)
            print(root.data, end = " ")
            preorder(root.right)
```

```python
ineorder(root)
```

```python
def posteorder(root):
        if root != None:
            preorder(root.left)
            preorder(root.right)
            print(root.data, end = " ")
```

```python
posteorder(root)
```
