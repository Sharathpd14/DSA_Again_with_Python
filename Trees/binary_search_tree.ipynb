```python
class TreeNode:
    def __init__(self,value):
        self.data = value
        self.left = None
        self.right = None
    
def insert(self, value):
    if self == None:
        return TreeNode(value)
    
    if self.data == value:
        return self
    
    if self.data > value:
        self.left = insert(self.left, value)
    else:
        self.right = insert(self.right,value)
        
    return self
        
    
```

```python
def search(self, value):
    if self == None:
        print("Element not found")
        return 
    
    if self.data == value:
        print("Element found")
        return
            
    if self.data > value:
        search(self.left, value)
    else:
        search(self.right,value)
    
```

```python
def inorder(self):
    if self != None:
        inorder(self.left)
        print(self.data, end = " ")
        inorder(self.right)
```
