```python
class TreeNode:
    def __init__(self,value):
        self.data = value
        self.left = None
        self.right = None
    
def insert(self, value):
    if self == None:
        return TreeNode(value)
    
    if self.data == value:
        return self
    
    if self.data > value:
        self.left = insert(self.left, value)
    else:
        self.right = insert(self.right,value)
        
    return self
        
    
```

```python
def get_successor(root):
    root = root.right
    
    while (root != None and root.left != None):
        root = root.left
    return root
```

```python
def search(self, value):
    if self == None:
        print("Element not found")
        return 
    
    if self.data == value:
        print("Element found")
        return
            
    if self.data > value:
        search(self.left, value)
    else:
        search(self.right,value)
    
```

```python
def delete(root,value):
    if root == None:
        print("Tree is empty")
        return root
    if root.data > value:
        root.left = delete(root.left, value)
    elif root.data < value:
        root.right = delete(root.right, value)
    else:
        if root.left == None:
            return root.right
        
        if root.right == None:
            return root.left
        else:
            successor = get_successor(root)
            root.data = successor.data
            root.right = delete(successor, successor.data)
            
    return root
        
```

```python
def inorder(self):
    if self != None:
        inorder(self.left)
        print(self.data, end = " ")
        inorder(self.right)
```
