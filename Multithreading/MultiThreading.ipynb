### Multi-threading

```python
import threading
import time
import sys
```

```python
def number_generator():
    for i in range(1, 11):
        time.sleep(1) 
        print(f"Generated number: {i}")

def letter_generator():
    for letter in 'abcdefghij':
        time.sleep(1)
        print(f"Generated letter: {letter}")
        
```

```python
t1 = threading.Thread(target=number_generator)
t2 = threading.Thread(target=letter_generator)
```

```python
t = time.perf_counter()
t1.start()
t2.start()

t1.join()
t2.join()
fi_t = time.perf_counter() - t
print(f"Time taken without threading: {fi_t:.12f} seconds")
```
